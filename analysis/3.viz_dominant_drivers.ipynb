```sql
CREATE OR REPLACE TEMPORARY VIEW dominant_drives AS
SELECT driver_name,
       COUNT(1) AS total_races,
       SUM(calculated_points) AS total_points,
       AVG(calculated_points) AS avg_points,
       RANK() OVER (ORDER BY AVG(calculated_points) DESC) AS rank
  FROM f1_presentation.calculated_race_results
GROUP BY driver_name
HAVING COUNT(1) >= 50
ORDER BY avg_points DESC;
```

```sql
SELECT race_year,
       driver_name,
       COUNT(1) AS total_races,
       SUM(calculated_points) AS total_points,
       AVG(calculated_points) AS avg_points
  FROM f1_presentation.calculated_race_results
  WHERE driver_name IN (SELECT driver_name FROM dominant_drives WHERE rank <= 10)
GROUP BY driver_name, race_year
ORDER BY avg_points DESC;
```
